```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import os

base_dir = os.path.dirname(os.path.abspath(__file__)) if "__file__" in globals() else os.getcwd()

file_path = os.path.join(base_dir, "PixelMisclassificationResults.csv")
save_path = os.path.join(base_dir, "Figure6.png")

df = pd.read_csv(file_path)

df = df.rename(columns={"Pixel Misclassification": "Class"})

df = df[~df["Class"].isin([
    "Number of pixels misclassified",
    "Number of pixels no data"
])]

df = df.reset_index(drop=True)

x = np.arange(len(df))
width = 0.35

fig, ax = plt.subplots(figsize=(10, 6))

ax.bar(
    x - width / 2,
    df["Misclassified pixels before additional features"],
    width,
    label="Misclassified pixels before additional features",
    color="orange"
)

ax.bar(
    x + width / 2,
    df["Misclassified pixels after additional features"],
    width,
    label="Misclassified pixels after additional features",
    color="steelblue"
)

ax.set_xlabel("Class")
ax.set_ylabel("Number of misclassified pixels")
ax.set_xticks(x)
ax.set_xticklabels(df["Class"])
ax.legend()

plt.tight_layout()

fig.savefig(save_path, dpi=300, bbox_inches="tight")
print(f"Figure saved to: {save_path}")

plt.show()
plt.close(fig)
```
